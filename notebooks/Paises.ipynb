```python
import pandas as pd
import plotly.express as px
from IPython.display import display, HTML

# Criar DataFrame (simulando a planilha Excel)
dados = {
    "País": [
        "Brasil", "Estados Unidos", "China", "Índia", "Alemanha",
        "Japão", "França", "Canadá", "Austrália", "África do Sul"
    ],
    "Continente": [
        "América do Sul", "América do Norte", "Ásia", "Ásia", "Europa",
        "Ásia", "Europa", "América do Norte", "Oceania", "África"
    ],
    "População (milhões)": [214, 332, 1412, 1408, 83, 125, 65, 39, 26, 61],
    "PIB (US$ trilhões)": [2.3, 26.9, 17.7, 3.7, 4.4, 4.2, 3.1, 2.2, 1.8, 0.4],
    "Expectativa de Vida": [75, 78, 77, 70, 81, 84, 82, 82, 83, 64],
    "IDH": [0.758, 0.926, 0.768, 0.644, 0.942, 0.925, 0.901, 0.936, 0.951, 0.713],
}

df = pd.DataFrame(dados)

# ================================================
# 1️⃣ Gráfico — Mapa de População
# ================================================
fig_mapa = px.choropleth(
    df,
    locations="País",
    locationmode="country names",
    color="População (milhões)",
    hover_name="País",
    title="🌍 População por País (em milhões)",
    color_continuous_scale="viridis"
)
fig_mapa.show()

# ================================================
# 2️⃣ Gráfico — Ranking de PIB
# ================================================
fig_pib = px.bar(
    df.sort_values("PIB (US$ trilhões)", ascending=False),
    x="País",
    y="PIB (US$ trilhões)",
    color="PIB (US$ trilhões)",
    title="💰 Ranking de PIB (em trilhões de US$)",
    text_auto=True,
    color_continuous_scale="plasma"
)
fig_pib.show()

# ================================================
# 3️⃣ Gráfico — IDH x Expectativa de Vida
# ================================================
fig_idh = px.scatter(
    df,
    x="IDH",
    y="Expectativa de Vida",
    size="População (milhões)",
    color="Continente",
    hover_name="País",
    title="📈 IDH x Expectativa de Vida",
)
fig_idh.show()

# ================================================
# 4️⃣ Tabela de Ranking Populacional
# ================================================
df_rank = df.sort_values("População (milhões)", ascending=False).reset_index(drop=True)
df_rank.index += 1
display(HTML("<h3>🏆 Ranking Populacional</h3>"))
display(df_rank[["País", "População (milhões)", "PIB (US$ trilhões)", "IDH"]])
```
